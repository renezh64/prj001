```python
import sys
from pathlib import Path

# 1. Directorio de trabajo actual
directorio_actual = Path().resolve()
print("=" * 60)
print(f"📍 DIRECTORIO ACTUAL: {directorio_actual}")
print("=" * 60)

# 2. Archivos en la carpeta actual
print("\n📁 Archivos en el directorio actual:")
for elemento in directorio_actual.iterdir():
    tipo = "[Carpeta]" if elemento.is_dir() else "[Archivo]"
    print(f"  {tipo} {elemento.name}")

# 3. Archivos en la carpeta padre (debería ser 'src')
directorio_padre = directorio_actual.parent
print(f"\n📁 Archivos en la carpeta padre ({directorio_padre.name}):")
for elemento in directorio_padre.iterdir():
    tipo = "[Carpeta]" if elemento.is_dir() else "[Archivo]"
    print(f"  {tipo} {elemento.name}")

# 4. Verificar contenido exacto dentro de 'clases_po'
ruta_clases_po = directorio_actual / "clases_po"
print(f"\n🔍 Contenido dentro de 'clases_po':")
if ruta_clases_po.exists():
    for elemento in ruta_clases_po.iterdir():
        print(f"  - {elemento.name}")
else:
    print("  ⚠️ No se encontró la carpeta 'clases_po' en la carpeta padre.")

# 5. Rutas donde Python busca módulos
print("\n🐍 sys.path (primeras 3 rutas donde Python busca módulos):")
for path in sys.path[:3]:
    print(f"  - {path}")
print("=" * 60)
```
